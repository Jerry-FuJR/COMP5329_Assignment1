# Stage 3 Experiment: Attention Head Count

This notebook compares `num_heads = 1`, `2`, `4`, `8`, and `16` under the shared baseline.

## Research Question

Under a fixed model width (`d_model = 96`), how does the number of attention heads affect optimization behaviour and QA performance in the repaired QANet?

## Hypothesis

A moderate number of heads will outperform both the single-head setting and very narrow heads, because it better balances per-head capacity and subspace diversity.

## Shared Baseline

Unless otherwise stated, all settings follow the shared baseline in `baseline参数.md`.

Only the following variable is changed in this experiment:

- `num_heads`

The notebook is organised into three parts:

1. Train all head-count settings.
2. Run a unified evaluation pass on all saved checkpoints.
3. Summarise and visualise the recorded metrics.

---
## Train All Settings

```python
from TrainTools.train import train
from EvaluateTools.evaluate import evaluate

import matplotlib.pyplot as plt
import pandas as pd

HEAD_SETTINGS = [1, 2, 4, 8, 16]

train_results = {}
eval_results = {}
```

---
## Run Training

```python
for heads in HEAD_SETTINGS:
    print(f"===== Training num_heads={heads} =====")
    train_results[heads] = train(
        train_npz       = "_data/train.npz",
        dev_npz         = "_data/dev.npz",
        word_emb_json   = "_data/word_emb.json",
        char_emb_json   = "_data/char_emb.json",
        train_eval_json = "_data/train_eval.json",
        dev_eval_json   = "_data/dev_eval.json",

        save_dir   = f"experiments/_model_heads_{heads}",
        log_dir    = f"experiments/_log_heads_{heads}",

        num_steps  = 1000,
        num_heads  = heads,

        optimizer_name = "sgd",
        scheduler_name = "lambda",
        loss_name      = "qa_nll",
    )
    print(
        f"Heads={heads} | Best F1: {train_results[heads]['best_f1']:.4f} | "
        f"Best EM: {train_results[heads]['best_em']:.4f}"
    )
```

---
## Unified Evaluation

```python
for heads in HEAD_SETTINGS:
    print(f"===== Evaluating num_heads={heads} =====")
    eval_results[heads] = evaluate(
        save_dir  = f"experiments/_model_heads_{heads}",
        log_dir   = f"experiments/_log_heads_{heads}",
        num_heads = heads,
    )
    print(
        f"Heads={heads} | Eval F1: {eval_results[heads]['f1']:.4f} | "
        f"Eval EM: {eval_results[heads]['exact_match']:.4f} | "
        f"Eval Loss: {eval_results[heads]['loss']:.6f}"
    )
```

---
## Summary Table and Visual Comparison

The first table combines checkpoint-based training metrics and final evaluation metrics. The figures then compare convergence behaviour and final performance.

```python
summary_rows = []

for heads in HEAD_SETTINGS:
    history = train_results[heads]["history"]
    last_entry = history[-1]
    best_entry = max(history, key=lambda row: row["dev_f1"])

    summary_rows.append({
        "num_heads": heads,
        "d_k": 96 // heads,
        "last_step": last_entry["step"],
        "final_train_loss": last_entry["train_loss"],
        "final_dev_loss": last_entry["dev_loss"],
        "best_dev_f1": train_results[heads]["best_f1"],
        "best_dev_em": train_results[heads]["best_em"],
        "best_dev_step": best_entry["step"],
        "eval_f1": eval_results[heads]["f1"],
        "eval_em": eval_results[heads]["exact_match"],
        "eval_loss": eval_results[heads]["loss"],
    })

summary_df = pd.DataFrame(summary_rows).sort_values("num_heads").reset_index(drop=True)
summary_df
```

---
## Convergence Curves

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=True)

for heads in HEAD_SETTINGS:
    history_df = pd.DataFrame(train_results[heads]["history"])
    label = f"heads={heads} (d_k={96 // heads})"

    axes[0, 0].plot(history_df["step"], history_df["train_loss"], marker="o", label=label)
    axes[0, 1].plot(history_df["step"], history_df["dev_loss"], marker="o", label=label)
    axes[1, 0].plot(history_df["step"], history_df["dev_f1"], marker="o", label=label)
    axes[1, 1].plot(history_df["step"], history_df["dev_em"], marker="o", label=label)

axes[0, 0].set_title("Train Loss")
axes[0, 1].set_title("Dev Loss")
axes[1, 0].set_title("Dev F1")
axes[1, 1].set_title("Dev EM")

for ax in axes.flat:
    ax.set_xlabel("Training Step")
    ax.grid(True, alpha=0.3)
    ax.legend()

plt.tight_layout()
plt.show()
```

---
## Final Metric Comparison

```python
plot_df = summary_df.copy()
x_labels = [str(h) for h in plot_df["num_heads"]]

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

axes[0].bar(x_labels, plot_df["best_dev_f1"], color="#4C78A8")
axes[0].set_title("Best Dev F1")
axes[0].set_xlabel("num_heads")

axes[1].bar(x_labels, plot_df["best_dev_em"], color="#F58518")
axes[1].set_title("Best Dev EM")
axes[1].set_xlabel("num_heads")

axes[2].bar(x_labels, plot_df["eval_loss"], color="#54A24B")
axes[2].set_title("Final Eval Loss")
axes[2].set_xlabel("num_heads")

for ax in axes:
    ax.grid(True, axis="y", alpha=0.3)

plt.tight_layout()
plt.show()


fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(x_labels, plot_df["eval_f1"], color="#E45756")
axes[0].set_title("Final Eval F1")
axes[0].set_xlabel("num_heads")

axes[1].bar(x_labels, plot_df["eval_em"], color="#72B7B2")
axes[1].set_title("Final Eval EM")
axes[1].set_xlabel("num_heads")

for ax in axes:
    ax.grid(True, axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
def fmt_metric(value, digits=4):
    if pd.isna(value):
        return "N/A"
    return f"{value:.{digits}f}"

for _, row in summary_df.iterrows():
    print(
        f"Heads={int(row['num_heads'])} | d_k={int(row['d_k'])} | "
        f"Best Dev F1={fmt_metric(row['best_dev_f1'])} | Best Dev EM={fmt_metric(row['best_dev_em'])} | "
        f"Eval F1={fmt_metric(row['eval_f1'])} | Eval EM={fmt_metric(row['eval_em'])} | "
        f"Eval Loss={fmt_metric(row['eval_loss'], digits=6)}"
    )
```
